```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
claim = "The Eiffel Tower is in London"
fact = "The Eiffel Tower is in Paris"

claim_vec = model.encode([claim])
fact_vec = model.encode([fact])

score = cosine_similarity(claim_vec, fact_vec)
```

```python
print(score[0][0])
```

```python
score = score[0][0]

if score > 0.7:
    print("SUPPORTED")
elif score < 0.3:
    print("HALLUCINATION")
else:
    print("UNCERTAIN")
```
