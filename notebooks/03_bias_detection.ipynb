```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
texts = [
    "Women are emotional",
    "Men are natural leaders",
    "Girls are weak in sports",
    "Everyone deserves equal opportunities",
    "Men don't cry"
]
```

```python
bias_phrases = [
    "women are emotional",
    "men are natural leaders",
    "girls are weak",
    "men don't cry"
]
```

```python
results = []

for text in texts:

    text_lower = text.lower()

    bias_found = False

    detected_phrase = None

    for phrase in bias_phrases:

        if phrase in text_lower:

            bias_found = True
            detected_phrase = phrase

    results.append({
        "Text": text,
        "Bias Detected": bias_found,
        "Detected Phrase": detected_phrase
    })
```

```python
df = pd.DataFrame(results)

print(df)
```

```python
bias_scores = []

for item in results:

    if item["Bias Detected"]:
        bias_scores.append(1)
    else:
        bias_scores.append(0)

df["Bias Score"] = bias_scores
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Bias Detected",
    data=df
)

plt.title("Bias Detection Distribution")

plt.xlabel("Bias Found")

plt.ylabel("Count")

plt.show()
```

```python
df["Bias Detected"].value_counts().plot.pie(
    autopct='%1.1f%%',
    figsize=(6,6)
)

plt.title("Bias Percentage Distribution")

plt.ylabel("")

plt.show()
```
